```python
import pandas as pd
from pathlib import Path

print("FINAL BUSINESS INSIGHTS")
print("=" * 50)

PROJECT_ROOT = Path.cwd().parent

RISK_PATH = PROJECT_ROOT / "reports" / "risk_scoring.csv"

df = pd.read_csv(RISK_PATH, low_memory=False)

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nRisk Level Distribution:")
print(df["Risk_Level"].value_counts())

print("\nRisk Level Percentage:")
print(
    df["Risk_Level"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nAverage Units Sold:", round(df["units_sold"].mean(), 2))
print("Average Predicted Units Sold:", round(df["Predicted_Units_Sold"].mean(), 2))

print("\nAverage Risk Score:", round(df["Risk_Score"].mean(), 2))

print("\nAverage Inventory Gap:", round(df["Inventory_Gap"].mean(), 2))

print("\nAverage Inventory Coverage:", round(df["Inventory_Coverage"].mean(), 2))
```

```python
print("HIGHEST-RISK SKU ANALYSIS")
print("=" * 50)

sku_risk = (
    df.groupby("sku_id")
    .agg(
        Average_Risk_Score=("Risk_Score", "mean"),
        High_Risk_Rows=("Risk_Level", lambda x: (x == "High").sum()),
        Average_Inventory_Coverage=("Inventory_Coverage", "mean"),
        Average_Inventory_Gap=("Inventory_Gap", "mean"),
        Average_Units_Sold=("units_sold", "mean"),
        Average_Predicted_Units=("Predicted_Units_Sold", "mean")
    )
    .sort_values(
        ["Average_Risk_Score", "High_Risk_Rows"],
        ascending=False
    )
)

print("Top 20 highest-risk SKUs:")
print(sku_risk.head(20).round(2).to_string())

print("\nTotal SKUs:", len(sku_risk))

print("\nSKUs with High Risk rows:")
print((sku_risk["High_Risk_Rows"] > 0).sum())
```

```python
# Step 3: Category and Subcategory Business Analysis

print("CATEGORY & SUBCATEGORY BUSINESS ANALYSIS")
print("=" * 50)

category_analysis = (
    df.groupby("category")
    .agg(
        Average_Risk_Score=("Risk_Score", "mean"),
        High_Risk_Rows=("Risk_Level", lambda x: (x == "High").sum()),
        Average_Inventory_Coverage=("Inventory_Coverage", "mean"),
        Average_Inventory_Gap=("Inventory_Gap", "mean"),
        Average_Units_Sold=("units_sold", "mean"),
        Average_Predicted_Units=("Predicted_Units_Sold", "mean"),
        Total_Revenue=("revenue", "sum")
    )
    .sort_values("Average_Risk_Score", ascending=False)
)

print("Category Analysis:")
print(category_analysis.round(2).to_string())

print("\nSubcategory Analysis:")

subcategory_analysis = (
    df.groupby("subcategory")
    .agg(
        Average_Risk_Score=("Risk_Score", "mean"),
        High_Risk_Rows=("Risk_Level", lambda x: (x == "High").sum()),
        Average_Inventory_Coverage=("Inventory_Coverage", "mean"),
        Average_Inventory_Gap=("Inventory_Gap", "mean"),
        Average_Units_Sold=("units_sold", "mean"),
        Total_Revenue=("revenue", "sum")
    )
    .sort_values("Average_Risk_Score", ascending=False)
)

print(subcategory_analysis.round(2).to_string())
```

```python
# Step 4: Time-Based Business Analysis

print("TIME-BASED BUSINESS ANALYSIS")
print("=" * 50)

df["date"] = pd.to_datetime(df["date"])

monthly_analysis = (
    df.groupby("month")
    .agg(
        Average_Risk_Score=("Risk_Score", "mean"),
        High_Risk_Rows=("Risk_Level", lambda x: (x == "High").sum()),
        Average_Inventory_Coverage=("Inventory_Coverage", "mean"),
        Average_Inventory_Gap=("Inventory_Gap", "mean"),
        Average_Units_Sold=("units_sold", "mean"),
        Average_Predicted_Units=("Predicted_Units_Sold", "mean"),
        Total_Revenue=("revenue", "sum")
    )
    .sort_index()
)

print("Monthly Analysis:")
print(monthly_analysis.round(2).to_string())

print("\nSeason Analysis:")

season_analysis = (
    df.groupby("season")
    .agg(
        Average_Risk_Score=("Risk_Score", "mean"),
        High_Risk_Rows=("Risk_Level", lambda x: (x == "High").sum()),
        Average_Inventory_Coverage=("Inventory_Coverage", "mean"),
        Average_Inventory_Gap=("Inventory_Gap", "mean"),
        Average_Units_Sold=("units_sold", "mean"),
        Total_Revenue=("revenue", "sum")
    )
    .sort_values("Average_Risk_Score", ascending=False)
)

print(season_analysis.round(2).to_string())
```

```python
# Step 5: Promotion and Holiday Business Analysis

print("PROMOTION & HOLIDAY BUSINESS ANALYSIS")
print("=" * 50)

promo_analysis = (
    df.groupby("promo_flag")
    .agg(
        Average_Risk_Score=("Risk_Score", "mean"),
        High_Risk_Rows=("Risk_Level", lambda x: (x == "High").sum()),
        Average_Inventory_Coverage=("Inventory_Coverage", "mean"),
        Average_Inventory_Gap=("Inventory_Gap", "mean"),
        Average_Units_Sold=("units_sold", "mean"),
        Average_Predicted_Units=("Predicted_Units_Sold", "mean"),
        Total_Revenue=("revenue", "sum")
    )
)

print("Promotion Analysis:")
print(promo_analysis.round(2).to_string())

print("\nHoliday Analysis:")

holiday_analysis = (
    df.groupby("is_holiday")
    .agg(
        Average_Risk_Score=("Risk_Score", "mean"),
        High_Risk_Rows=("Risk_Level", lambda x: (x == "High").sum()),
        Average_Inventory_Coverage=("Inventory_Coverage", "mean"),
        Average_Inventory_Gap=("Inventory_Gap", "mean"),
        Average_Units_Sold=("units_sold", "mean"),
        Average_Predicted_Units=("Predicted_Units_Sold", "mean"),
        Total_Revenue=("revenue", "sum")
    )
)

print(holiday_analysis.round(2).to_string())

print("\nPromo Event Analysis:")

promo_event_analysis = (
    df.groupby("promo_event")
    .agg(
        Average_Risk_Score=("Risk_Score", "mean"),
        High_Risk_Rows=("Risk_Level", lambda x: (x == "High").sum()),
        Average_Inventory_Coverage=("Inventory_Coverage", "mean"),
        Average_Units_Sold=("units_sold", "mean"),
        Total_Revenue=("revenue", "sum")
    )
    .sort_values("Average_Risk_Score", ascending=False)
)

print(promo_event_analysis.round(2).to_string())
```

```python
# Step 6: Final Business Insight Summary

print("FINAL BUSINESS INSIGHT SUMMARY")
print("=" * 50)

high_risk_pct = (df["Risk_Level"] == "High").mean() * 100
medium_risk_pct = (df["Risk_Level"] == "Medium").mean() * 100
low_risk_pct = (df["Risk_Level"] == "Low").mean() * 100

promo_units = df.loc[df["promo_flag"] == 1, "units_sold"].mean()
non_promo_units = df.loc[df["promo_flag"] == 0, "units_sold"].mean()

promo_coverage = df.loc[df["promo_flag"] == 1, "Inventory_Coverage"].mean()
non_promo_coverage = df.loc[df["promo_flag"] == 0, "Inventory_Coverage"].mean()

holiday_coverage = df.loc[df["is_holiday"] == 1, "Inventory_Coverage"].mean()
non_holiday_coverage = df.loc[df["is_holiday"] == 0, "Inventory_Coverage"].mean()

highest_risk_sku = sku_risk.index[0]
highest_risk_sku_score = sku_risk.iloc[0]["Average_Risk_Score"]

highest_risk_category = category_analysis.index[0]
highest_risk_category_score = category_analysis.iloc[0]["Average_Risk_Score"]

highest_risk_month = monthly_analysis["Average_Risk_Score"].idxmax()
highest_risk_month_score = monthly_analysis["Average_Risk_Score"].max()

highest_risk_season = season_analysis.index[0]
highest_risk_season_score = season_analysis.iloc[0]["Average_Risk_Score"]

print("Risk Distribution:")
print(f"High Risk: {high_risk_pct:.2f}%")
print(f"Medium Risk: {medium_risk_pct:.2f}%")
print(f"Low Risk: {low_risk_pct:.2f}%")

print("\nKey Inventory Findings:")
print(f"Average Risk Score: {df['Risk_Score'].mean():.2f}")
print(f"Average Inventory Coverage: {df['Inventory_Coverage'].mean():.2f}")
print(f"Average Inventory Gap: {df['Inventory_Gap'].mean():.2f}")

print("\nHighest-Risk SKU:")
print(f"{highest_risk_sku}: {highest_risk_sku_score:.2f}")

print("\nHighest-Risk Category:")
print(f"{highest_risk_category}: {highest_risk_category_score:.2f}")

print("\nHighest-Risk Month:")
print(f"Month {highest_risk_month}: {highest_risk_month_score:.2f}")

print("\nHighest-Risk Season:")
print(f"{highest_risk_season}: {highest_risk_season_score:.2f}")

print("\nPromotion Impact:")
print(f"Promotional average units sold: {promo_units:.2f}")
print(f"Non-promotional average units sold: {non_promo_units:.2f}")
print(f"Promotional inventory coverage: {promo_coverage:.2f}")
print(f"Non-promotional inventory coverage: {non_promo_coverage:.2f}")

print("\nHoliday Impact:")
print(f"Holiday inventory coverage: {holiday_coverage:.2f}")
print(f"Non-holiday inventory coverage: {non_holiday_coverage:.2f}")

print("\nFinal business insight analysis completed.")
```
