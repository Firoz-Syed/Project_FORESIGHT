```python
import pandas as pd

sales = pd.read_csv("../data/raw/sales_daily.csv")
sku = pd.read_csv("../data/raw/sku_master.csv")
calendar = pd.read_csv("../data/raw/calendar.csv")
inventory = pd.read_csv("../data/raw/inventory_snapshots.csv")
```

```python
master = pd.merge(
    sales,
    sku,
    on="sku_id",
    how="left"
)
```

```python
master = pd.merge(
    master,
    calendar,
    on="date",
    how="left"
)
```

```python
master = pd.merge(
    master,
    inventory,
    on=["date", "sku_id"],
    how="left"
)
```
