```python
import pandas as pd
import numpy as np
from pathlib import Path

print("Final Data Validation Started")
print("=" * 50)

# Project root
PROJECT_ROOT = Path("..")

# Final processed dataset
DATA_PATH = PROJECT_ROOT / "data" / "processed" / "feature_engineered_dataset.csv"

print("Dataset path:")
print(DATA_PATH.resolve())

# Check file exists
if not DATA_PATH.exists():
    print("\n ERROR: Dataset file not found!")
else:
    print("\n Dataset file found!")
```

```python
# Step 2: Load and inspect the final dataset

df = pd.read_csv(DATA_PATH)

print("Dataset Loaded Successfully")
print("=" * 50)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# Step 3: Missing Values Validation

missing_values = df.isnull().sum()

print("Missing Values Validation")
print("=" * 50)

print(missing_values)

total_missing = missing_values.sum()

print("\nTotal Missing Values:", total_missing)

if total_missing == 0:
    print(" PASS: No missing values found.")
else:
    print(" WARNING: Missing values found.")
```

```python
# Step 4: Detailed Missing Value Analysis

print("Detailed Missing Value Analysis")
print("=" * 50)

missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df) * 100).round(2)
})

missing_summary = missing_summary[missing_summary["Missing_Count"] > 0]

print(missing_summary)
```

```python
# Step 5: Investigate Missing Inventory Data

print("Missing Inventory Data Investigation")
print("=" * 50)

inventory_cols = [
    "on_hand_units",
    "on_order_units",
    "lead_time_days",
    "reorder_point"
]

print("\nMissing values by column:")
print(df[inventory_cols].isnull().sum())

print("\nRows where inventory data is missing:")
missing_inventory = df[df[inventory_cols].isnull().any(axis=1)]

print("Number of affected rows:", len(missing_inventory))

print("\nSample affected rows:")
display(
    missing_inventory[
        ["date", "sku_id"] + inventory_cols
    ].head(10)
)
```

```python
# Step 6: Check Missing Inventory Pattern

print("Missing Inventory Pattern")
print("=" * 50)

print("\nDate range of missing inventory rows:")
print("Start:", missing_inventory["date"].min())
print("End:", missing_inventory["date"].max())

print("\nNumber of unique SKUs affected:")
print(missing_inventory["sku_id"].nunique())

print("\nTop affected SKUs:")
print(missing_inventory["sku_id"].value_counts().head(10))

print("\nMissing inventory rows by date:")
print(missing_inventory["date"].value_counts().head(10))
```

```python
# Step 7: Validate Original Inventory Data

inventory_path = PROJECT_ROOT / "data" / "raw" / "inventory_snapshots.csv"

print("Original Inventory Data Validation")
print("=" * 50)

print("Inventory file path:")
print(inventory_path.resolve())

if inventory_path.exists():
    inventory_df = pd.read_csv(inventory_path)

    print("\n Original inventory file found!")
    print("Rows:", inventory_df.shape[0])
    print("Columns:", inventory_df.shape[1])

    print("\nColumns:")
    print(inventory_df.columns.tolist())

    print("\nMissing values in original inventory data:")
    print(inventory_df.isnull().sum())

else:
    print("\n Original inventory file NOT found!")
```

```python
# Step 8: Compare Inventory Coverage

print("Inventory Coverage Check")
print("=" * 50)

print("\nFinal Dataset:")
print("Date range:", df["date"].min(), "to", df["date"].max())
print("Unique SKUs:", df["sku_id"].nunique())

print("\nOriginal Inventory:")
print("Date range:", inventory_df["date"].min(), "to", inventory_df["date"].max())
print("Unique SKUs:", inventory_df["sku_id"].nunique())

print("\nOriginal Inventory Rows:", len(inventory_df))

print("\nExpected inventory combinations:")
print(
    inventory_df["date"].nunique()
    * inventory_df["sku_id"].nunique()
)
```

```python
# Step 9: Check Date + SKU Matching

print("Date + SKU Matching Validation")
print("=" * 50)

# Make sure date types match
df["date"] = pd.to_datetime(df["date"])
inventory_df["date"] = pd.to_datetime(inventory_df["date"])

# Create matching keys
final_keys = set(zip(df["date"], df["sku_id"]))
inventory_keys = set(zip(inventory_df["date"], inventory_df["sku_id"]))

matched_keys = final_keys.intersection(inventory_keys)

print("Unique Date + SKU combinations in final dataset:", len(final_keys))
print("Unique Date + SKU combinations in inventory:", len(inventory_keys))
print("Matching Date + SKU combinations:", len(matched_keys))
print("Non-matching combinations:", len(final_keys - inventory_keys))

# Check actual inventory rows that should match
match_mask = [
    (date, sku) in inventory_keys
    for date, sku in zip(df["date"], df["sku_id"])
]

print("\nFinal dataset rows with matching inventory:")
print(sum(match_mask))

print("Final dataset rows without matching inventory:")
print(len(df) - sum(match_mask))
```

```python
# Step 10: Inventory Date Coverage

print("Inventory Date Coverage")
print("=" * 50)

inventory_date_counts = inventory_df.groupby("date").size()

print("Number of inventory dates:", len(inventory_date_counts))

print("\nFirst 10 inventory dates:")
print(inventory_date_counts.head(10))

print("\nLast 10 inventory dates:")
print(inventory_date_counts.tail(10))

print("\nRows per inventory date:")
print(inventory_date_counts.value_counts().sort_index())
```

```python
# Step 11: Display All Inventory Snapshot Dates

print("All Inventory Snapshot Dates")
print("=" * 50)

snapshot_dates = sorted(inventory_df["date"].unique())

for i, date in enumerate(snapshot_dates, start=1):
    print(f"{i:02d}. {pd.Timestamp(date).date()}")
```

```python
# Step 12: Test As-Of Inventory Coverage - Corrected

print("As-Of Inventory Coverage Test")
print("=" * 50)

# Prepare daily data
daily_df = df[["date", "sku_id"]].copy()

# Prepare inventory data
inventory_test = inventory_df[
    [
        "date",
        "sku_id",
        "on_hand_units",
        "on_order_units",
        "lead_time_days",
        "reorder_point"
    ]
].copy()

# Make sure dates are datetime
daily_df["date"] = pd.to_datetime(daily_df["date"])
inventory_test["date"] = pd.to_datetime(inventory_test["date"])

# Sort by date first, then SKU
daily_df = daily_df.sort_values(["date", "sku_id"]).reset_index(drop=True)
inventory_test = inventory_test.sort_values(["date", "sku_id"]).reset_index(drop=True)

# As-of matching
test_merged = pd.merge_asof(
    daily_df,
    inventory_test,
    on="date",
    by="sku_id",
    direction="backward"
)

inventory_cols = [
    "on_hand_units",
    "on_order_units",
    "lead_time_days",
    "reorder_point"
]

print("Rows tested:", len(test_merged))

print("\nMissing values after as-of matching:")
print(test_merged[inventory_cols].isnull().sum())

total_missing_after = test_merged[inventory_cols].isnull().sum().sum()

print("\nTotal inventory missing after as-of matching:", total_missing_after)

if total_missing_after == 0:
    print(" PASS: All daily rows can receive an inventory snapshot.")
else:
    print(" Some inventory values still cannot be matched.")
```

```python
# Step 13: Apply As-Of Inventory Data to Final Dataset

print("Applying As-Of Inventory Data")
print("=" * 50)

# Keep the original final dataset columns
feature_cols = [
    "on_hand_units",
    "on_order_units",
    "lead_time_days",
    "reorder_point"
]

# Remove the currently incorrect inventory columns
df_base = df.drop(columns=feature_cols)

# Prepare inventory data
inventory_final = inventory_df[
    ["date", "sku_id"] + feature_cols
].copy()

# Ensure datetime format
df_base["date"] = pd.to_datetime(df_base["date"])
inventory_final["date"] = pd.to_datetime(inventory_final["date"])

# Sort for merge_asof
df_base = df_base.sort_values(["date", "sku_id"]).reset_index(drop=True)
inventory_final = inventory_final.sort_values(["date", "sku_id"]).reset_index(drop=True)

# Apply latest available inventory snapshot
df_fixed = pd.merge_asof(
    df_base,
    inventory_final,
    on="date",
    by="sku_id",
    direction="backward"
)

print("Original rows:", len(df))
print("Fixed rows:", len(df_fixed))

print("\nMissing inventory values after correction:")
print(df_fixed[feature_cols].isnull().sum())

total_inventory_missing = df_fixed[feature_cols].isnull().sum().sum()

print("\nTotal inventory missing:", total_inventory_missing)

if total_inventory_missing == 0:
    print("PASS: Inventory data successfully applied to all rows.")
else:
    print("FAIL: Inventory values are still missing.")
```

```python
# Step 14: Investigate Promo Event Missing Values

print("Promo Event Validation")
print("=" * 50)

print("Missing promo_event values:")
print(df_fixed["promo_event"].isnull().sum())

print("\nNon-missing promo_event values:")
print(df_fixed["promo_event"].notnull().sum())

print("\nUnique promo_event values:")
print(df_fixed["promo_event"].value_counts(dropna=False))

print("\nPromo event data type:")
print(df_fixed["promo_event"].dtype)
```

```python
# Step 15: Validate Original Calendar Data

calendar_path = PROJECT_ROOT / "data" / "raw" / "calender.csv"

print("Original Calendar Data Validation")
print("=" * 50)

print("Calendar file path:")
print(calendar_path.resolve())

if calendar_path.exists():
    calendar_df = pd.read_csv(calendar_path)

    print("\nOriginal calendar file found!")

    print("Rows:", calendar_df.shape[0])
    print("Columns:", calendar_df.shape[1])

    print("\nColumns:")
    print(calendar_df.columns.tolist())

    print("\nMissing values:")
    print(calendar_df.isnull().sum())

    print("\nPromo event values:")
    print(calendar_df["promo_event"].value_counts(dropna=False))

else:
    print("\nOriginal calendar file NOT found!")
```

```python
# Step 16: Find Calendar File

raw_path = PROJECT_ROOT / "data" / "raw"

print("Files inside data/raw:")
print("=" * 50)

for file in raw_path.iterdir():
    print(file.name)
```

```python
# Step 17: Validate Original Calendar Data

calendar_path = PROJECT_ROOT / "data" / "raw" / "calendar.csv"

print("Original Calendar Data Validation")
print("=" * 50)

print("Calendar file path:")
print(calendar_path.resolve())

if calendar_path.exists():
    calendar_df = pd.read_csv(calendar_path)

    print("\nOriginal calendar file found!")

    print("Rows:", calendar_df.shape[0])
    print("Columns:", calendar_df.shape[1])

    print("\nColumns:")
    print(calendar_df.columns.tolist())

    print("\nMissing values:")
    print(calendar_df.isnull().sum())

    print("\nPromo event values:")
    print(calendar_df["promo_event"].value_counts(dropna=False))

else:
    print("\nOriginal calendar file NOT found!")
```

```python
# Step 18: Handle Missing Promo Events

print("Handling Promo Event Values")
print("=" * 50)

print("Before:")
print(df_fixed["promo_event"].value_counts(dropna=False))

# Replace missing promo events with explicit "No_Event"
df_fixed["promo_event"] = df_fixed["promo_event"].fillna("No_Event")

print("\nAfter:")
print(df_fixed["promo_event"].value_counts(dropna=False))

# Validate
missing_promo = df_fixed["promo_event"].isnull().sum()

print("\nMissing promo_event values:", missing_promo)

if missing_promo == 0:
    print("PASS: promo_event has no missing values.")
else:
    print("FAIL: promo_event still contains missing values.")
```

```python
# Step 19: Final Missing Value Validation

print("FINAL MISSING VALUE VALIDATION")
print("=" * 50)

missing_final = df_fixed.isnull().sum()

print(missing_final)

total_missing_final = missing_final.sum()

print("\nTotal Missing Values:", total_missing_final)

if total_missing_final == 0:
    print("\nPASS: Final dataset contains NO missing values.")
else:
    print("\nFAIL: Missing values still exist.")
    
    print("\nColumns with missing values:")
    print(missing_final[missing_final > 0])
```

```python
# Step 20: Duplicate Records Validation

print("DUPLICATE RECORDS VALIDATION")
print("=" * 50)

duplicate_count = df_fixed.duplicated().sum()

print("Total duplicate rows:", duplicate_count)

if duplicate_count == 0:
    print("\nPASS: No duplicate rows found.")
else:
    print("\nFAIL: Duplicate rows found.")
    
    print("\nSample duplicate rows:")
    display(df_fixed[df_fixed.duplicated(keep=False)].head(10))
```

```python
# Step 21: Data Type Validation

print("DATA TYPE VALIDATION")
print("=" * 50)

print("\nCurrent Data Types:")
print(df_fixed.dtypes)

# Expected datetime columns
datetime_columns = [
    "date",
    "launch_date"
]

# Expected numeric columns
numeric_columns = [
    "units_sold",
    "revenue",
    "unit_price",
    "promo_flag",
    "unit_cost",
    "list_price",
    "week",
    "month",
    "is_holiday",
    "on_hand_units",
    "on_order_units",
    "lead_time_days",
    "reorder_point",
    "day",
    "day_of_week",
    "quarter",
    "year"
]

print("\nDatetime Validation:")
datetime_errors = []

for col in datetime_columns:
    if not pd.api.types.is_datetime64_any_dtype(df_fixed[col]):
        datetime_errors.append(col)
        print(f"{col}: NOT datetime")
    else:
        print(f"{col}: datetime")

print("\nNumeric Validation:")
numeric_errors = []

for col in numeric_columns:
    if not pd.api.types.is_numeric_dtype(df_fixed[col]):
        numeric_errors.append(col)
        print(f"{col}: NOT numeric")
    else:
        print(f"{col}: numeric")

print("\nValidation Summary")

if len(datetime_errors) == 0 and len(numeric_errors) == 0:
    print("PASS: All expected data types are correct.")
else:
    print("FAIL: Some data types need correction.")
    print("Datetime errors:", datetime_errors)
    print("Numeric errors:", numeric_errors)
```

```python
# Step 22: Correct launch_date Data Type

print("Correcting launch_date Data Type")
print("=" * 50)

# Convert launch_date to datetime
df_fixed["launch_date"] = pd.to_datetime(
    df_fixed["launch_date"],
    errors="coerce"
)

print("launch_date dtype:", df_fixed["launch_date"].dtype)

# Check if conversion created missing values
invalid_launch_dates = df_fixed["launch_date"].isnull().sum()

print("Invalid/missing launch_date values:", invalid_launch_dates)

if invalid_launch_dates == 0:
    print("PASS: launch_date successfully converted to datetime.")
else:
    print("FAIL: Invalid launch_date values found.")
```

```python
# Step 23: Final Data Type Revalidation

print("FINAL DATA TYPE VALIDATION")
print("=" * 50)

datetime_columns = [
    "date",
    "launch_date"
]

numeric_columns = [
    "units_sold",
    "revenue",
    "unit_price",
    "promo_flag",
    "unit_cost",
    "list_price",
    "week",
    "month",
    "is_holiday",
    "on_hand_units",
    "on_order_units",
    "lead_time_days",
    "reorder_point",
    "day",
    "day_of_week",
    "quarter",
    "year"
]

datetime_errors = []
numeric_errors = []

print("\nDatetime columns:")

for col in datetime_columns:
    if pd.api.types.is_datetime64_any_dtype(df_fixed[col]):
        print(f"{col}")
    else:
        print(f"{col}")
        datetime_errors.append(col)

print("\nNumeric columns:")

for col in numeric_columns:
    if pd.api.types.is_numeric_dtype(df_fixed[col]):
        print(f"{col}")
    else:
        print(f"{col}")
        numeric_errors.append(col)

print("\nValidation Summary")

if not datetime_errors and not numeric_errors:
    print("PASS: All expected data types are correct.")
else:
    print("FAIL")
    print("Datetime errors:", datetime_errors)
    print("Numeric errors:", numeric_errors)
```

```python
# Step 24: Date Range Validation

print("DATE VALIDATION")
print("=" * 50)

print("Main date range:")
print("Start:", df_fixed["date"].min())
print("End:", df_fixed["date"].max())

print("\nLaunch date range:")
print("Start:", df_fixed["launch_date"].min())
print("End:", df_fixed["launch_date"].max())

print("\nInvalid date values:")

invalid_main_dates = df_fixed["date"].isna().sum()
invalid_launch_dates = df_fixed["launch_date"].isna().sum()

print("Invalid date:", invalid_main_dates)
print("Invalid launch_date:", invalid_launch_dates)

print("\nValidation Summary")

if invalid_main_dates == 0 and invalid_launch_dates == 0:
    print("PASS: All date values are valid.")
else:
    print("FAIL: Invalid date values found.")
```

```python
# Step 25: Numeric and Business Rule Validation

print("NUMERIC & BUSINESS-RULE VALIDATION")
print("=" * 50)

checks = {
    "units_sold_negative": (df_fixed["units_sold"] < 0).sum(),
    "revenue_negative": (df_fixed["revenue"] < 0).sum(),
    "unit_price_negative": (df_fixed["unit_price"] < 0).sum(),
    "unit_cost_negative": (df_fixed["unit_cost"] < 0).sum(),
    "list_price_negative": (df_fixed["list_price"] < 0).sum(),
    "on_hand_units_negative": (df_fixed["on_hand_units"] < 0).sum(),
    "on_order_units_negative": (df_fixed["on_order_units"] < 0).sum(),
    "lead_time_days_negative": (df_fixed["lead_time_days"] < 0).sum(),
    "reorder_point_negative": (df_fixed["reorder_point"] < 0).sum()
}

for check, count in checks.items():
    print(f"{check}: {count}")

print("\nValidation Summary")

if all(count == 0 for count in checks.values()):
    print("PASS: No invalid negative values found.")
else:
    print("WARNING: Some negative values were found.")
```

```python
# Step 26: Final Dataset Consistency Check

print("FINAL DATASET CONSISTENCY CHECK")
print("=" * 50)

expected_rows = 73000
expected_columns = 24
expected_skus = 200

actual_rows = len(df_fixed)
actual_columns = len(df_fixed.columns)
actual_skus = df_fixed["sku_id"].nunique()
unique_date_sku = df_fixed[["date", "sku_id"]].drop_duplicates().shape[0]

print("Rows:", actual_rows)
print("Columns:", actual_columns)
print("Unique SKUs:", actual_skus)
print("Unique Date + SKU combinations:", unique_date_sku)

print("\nExpected:")
print("Rows:", expected_rows)
print("Columns:", expected_columns)
print("SKUs:", expected_skus)
print("Date + SKU combinations:", expected_rows)

print("\nValidation Summary")

if (
    actual_rows == expected_rows
    and actual_columns == expected_columns
    and actual_skus == expected_skus
    and unique_date_sku == expected_rows
):
    print("PASS: Final dataset consistency is correct.")
else:
    print("FAIL: Final dataset consistency check failed.")
```
